### Task 06 (d)
Write a `normalize()` function and test it on messy inputs.

```python
# Task 06 (d)
def normalize(text):
    """Strip whitespace, lowercase, and remove commas and full-stops."""
    text = text.strip()
    text = text.lower()
    text = text.replace(",", "")
    text = text.replace(".", "")
    return text

# Test on messy inputs
messy1 = "  Hello, WORLD. This Is NLP!!!  "
messy2 = "   DATA Science, Is Fun.   "

print("Cleaned 1:", normalize(messy1))
print("Cleaned 2:", normalize(messy2))
```

Write a `validate_token()` function and apply it.

```python
# Task 07 (d)
def validate_token(token):
    """Return 'WORD' if alphabetic, 'NUMBER' if numeric, else 'MIXED'."""
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"

test_tokens = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]

for t in test_tokens:
    print(f"  '{t}' → {validate_token(t)}")
```

## Task 08: Functions and Lambda for NLP Pipelines
**Topic:** Functions, Lambda — Reusable Text Processing

### Task 08 (a)
Write a `clean()` function and an equivalent lambda.

```python
# Task 08 (a)
def clean(text):
    """Strip whitespace and convert to lowercase."""
    return text.strip().lower()

# Equivalent lambda
clean_lambda = lambda text: text.strip().lower()

test_str = "  HELLO Natural Language Processing  "
print("clean() output:", clean(test_str))
print("clean_lambda output:", clean_lambda(test_str))
```

### Task 08 (b)
Write `tokenize()` and `remove_short()` functions and chain them.

```python
# Task 08 (b)
def tokenize(text):
    """Split a cleaned string into a list of word tokens."""
    return text.split()

def remove_short(tokens, n):
    """Return only tokens whose length is greater than n."""
    return [t for t in tokens if len(t) > n]

sentence = "I love working with NLP tools in Python daily"
cleaned = clean(sentence)
tokens = tokenize(cleaned)
result = remove_short(tokens, 3)

print("Tokens after removing short words (len > 3):", result)
```

### Task 08 (c)
Write a `word_lengths()` function.

```python
# Task 08 (c)
def word_lengths(tokens):
    """Return a dict mapping each word to its character length."""
    return {word: len(word) for word in tokens}

tokens = ["language", "model", "ai", "processing", "deep"]
print("Word lengths:", word_lengths(tokens))
```

### Task 08 (d)
Build a full `nlp_pipeline()` that chains all previous functions.

```python
# Task 08 (d)
def nlp_pipeline(text, min_len):
    """Clean → Tokenize → Remove short words → Return length dict."""
    cleaned = clean(text)
    tokens = tokenize(cleaned)
    filtered = remove_short(tokens, min_len)
    return word_lengths(filtered)

# Test on two sentences
sent1 = "  Deep Learning is Revolutionizing NLP  "
sent2 = "  Transformers CHANGED the field of AI  "

print("Pipeline output 1:", nlp_pipeline(sent1, 3))
print("Pipeline output 2:", nlp_pipeline(sent2, 3))
```

## Task 09: map(), filter(), sorted() and Counter
**Topic:** `map()`, `filter()`, `sorted()`, Lambda, `collections.Counter`

### Task 09 (a)
Use `map()` with a lambda to lowercase and compute lengths.

```python
# Task 09 (a)
words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

# (i) Convert every word to lowercase
lower_words = list(map(lambda w: w.lower(), words))
print("Lowercase:", lower_words)

# (ii) Compute character length of each word
lengths = list(map(lambda w: len(w), words))
print("Lengths:", lengths)
```

### Task 09 (b)
Use `filter()` with a lambda to filter words.

```python
# Task 09 (b)
# (i) Words with more than 3 characters
words_long = list(filter(lambda w: len(w) > 3, words))
print("Words with > 3 chars:", words_long)

# (ii) Words containing the letter 't' (case-insensitive)
words_with_t = list(filter(lambda w: "t" in w.lower(), words))
print("Words containing 't':", words_with_t)
```

### Task 09 (c)
Sort the words list in three different ways.

```python
# Task 09 (c)
# (i) Alphabetically (default)
print("Alphabetical:", sorted(words))

# (ii) By word length ascending
print("By length (asc):", sorted(words, key=len))

# (iii) By word length descending
print("By length (desc):", sorted(words, key=len, reverse=True))
```

### Task 09 (d)
Use `Counter` for frequency analysis.

```python
# Task 09 (d)
from collections import Counter

sentence = "nlp nlp python ai nlp python text ai nlp python text"
word_tokens = sentence.split()

freq = Counter(word_tokens)
print("Full Counter:", freq)

# (i) 3 most common words
print("Top 3 most common:", freq.most_common(3))

# (ii) Frequency of "python"
print("Frequency of 'python':", freq["python"])

# (iii) Total words and unique vocabulary size
print("Total number of words:", sum(freq.values()))
print("Unique vocabulary size:", len(freq))
```
